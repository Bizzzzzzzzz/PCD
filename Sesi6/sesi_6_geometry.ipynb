```python
# Membaca gambar seperti biasa
img = cv2.imread("kucing.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

tinggi, lebar = img.shape[:2]

# ini proses translasi
M = np.float32([
    [1, 0, -150],
    [0, 1, 55]
])

# Melakukan translasi
hasil1 = cv2.warpAffine(
    img, M, (lebar, tinggi)
)

# Menampilkan hasil
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Citra Asli")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(hasil1)
plt.title("Hasil Translation")
plt.axis("off")

plt.show()
```

```python
img = cv2.imread("kucing.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

tinggi, lebar = img.shape[:2]

# Menentukan titik pusat
pusat = (lebar // 2, tinggi // 2)

# Rotasi 45 derajat
M1 = cv2.getRotationMatrix2D(
    pusat, 45, 1.0
)

M2 = cv2.getRotationMatrix2D(
    pusat, 90, 1.0
)

M3 = cv2.getRotationMatrix2D(
    pusat, 180, 1.0
)

hasil3 = cv2.warpAffine(
    img, M1, (lebar, tinggi)
)

hasil4 = cv2.warpAffine(
    img, M2, (lebar, tinggi)
)

hasil5 = cv2.warpAffine(
    img, M3, (lebar, tinggi)
)

plt.figure(figsize=(10, 5))

plt.subplot(2, 2, 1)
plt.imshow(img)
plt.title("Citra Asli")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(hasil3)
plt.title("Rotasi 45 Derajat")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(hasil4)
plt.title("Rotasi 90 Derajat")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(hasil5)
plt.title("Rotasi 180 Derajat")
plt.axis("off")

plt.show()
```

```python
# Membaca gambar
img = cv2.imread("kucing.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

hasil6 = cv2.flip(img, 1)

hasil7 = cv2.flip(img, 0)

plt.subplot(1, 3, 1)
plt.imshow(img)
plt.title(f"Gambar Asli")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(hasil6)
plt.title(f"Horizontal Flip")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(hasil7)
plt.title(f"Vertical Flip")
plt.axis("off")
```
